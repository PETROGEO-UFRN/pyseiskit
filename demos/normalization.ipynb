```python
import segyio
from bokeh.plotting import show, output_notebook
from bokeh.layouts import row
from seismicReader import readSeismicFile
from plotWiggles import plotWiggles
from pyseiskit import normalizeTraces
```

```python
# Load a raw shot gather (no TPOW applied)
gatherData, traceOffsets, timeSamples = readSeismicFile('tac-204RL239.su', gatherKey=segyio.TraceField.FieldRecord, gatherIndex=49)

# Apply Normalizations
maxNormalizedGather = normalizeTraces(gatherData, method='max')
rmsNormalizedGather = normalizeTraces(gatherData, method='rms')

# Plot side-by-side
originalPlotFigure = plotWiggles(gatherData, timeSamples, traceOffsets, "Raw Data", width=300)
maxNormalizedPlotFigure = plotWiggles(maxNormalizedGather, timeSamples, traceOffsets, "Max Normalized", width=300)
rmsNormalizedPlotFigure = plotWiggles(rmsNormalizedGather, timeSamples, traceOffsets, "RMS Normalized", width=300)

output_notebook()
show(row(originalPlotFigure, maxNormalizedPlotFigure, rmsNormalizedPlotFigure))
```
